```python
# [EXIGÊNCIA DE CÓDIGO 1 de 7]: Implementação do Nodo e da Lista Encadeada Simples
class Nodo:
    def __init__(self, numero, cor):
        self.numero = numero
        self.cor = cor
        self.proximo = None # [2]

class ListaEncadeada:
    def __init__(self):
        self.head = None # [2, 3]

    # [EXIGÊNCIA DE CÓDIGO 2 de 7]: Inserção no final da lista (Sem Prioridade)
    def inserirSemPrioridade(self, nodo):
        if self.head is None:
            self.head = nodo
        else:
            atual = self.head
            while atual.proximo is not None: # Anda pela lista até o último elemento [2]
                atual = atual.proximo
            atual.proximo = nodo # Insere no final [2]

    # [EXIGÊNCIA DE CÓDIGO 3 de 7]: Inserção após todos os "A" e antes dos "V"
    def inserirComPrioridade(self, nodo):
        # Caso a lista comece com um "V", o novo "A" vira a nova cabeça
        if self.head is None or self.head.cor == 'V':
            nodo.proximo = self.head
            self.head = nodo # [3]
        else:
            # Localiza o último nodo com cor "A" para inserção após ele
            atual = self.head
            while atual.proximo is not None and atual.proximo.cor == 'A':
                atual = atual.proximo
            nodo.proximo = atual.proximo
            atual.proximo = nodo # [2, 3]

# Variáveis globais para controle da numeração automática [3]
contador_v = 1
contador_a = 201
lista_hospital = ListaEncadeada()

# [EXIGÊNCIA DE CÓDIGO 4 de 7]: Função para solicitar dados e criar o nodo
def inserir():
    global contador_v, contador_a
    cor = input("Digite a cor do cartão (A ou V): ").strip().upper()

    if cor == 'V':
        numero = contador_v
        contador_v += 1
    elif cor == 'A':
        numero = contador_a
        contador_a += 1
    else:
        print("Cor inválida! Tente novamente.")
        return

    novo_nodo = Nodo(numero, cor) # [3]

    if lista_hospital.head is None:
        lista_hospital.head = novo_nodo # Se vazia, aponta o head para o nodo [3]
    else:
        if cor == 'V':
            lista_hospital.inserirSemPrioridade(novo_nodo) # [3]
        else:
            lista_hospital.inserirComPrioridade(novo_nodo) # [3]

# [EXIGÊNCIA DE CÓDIGO 5 de 7]: Impressão de todos os pacientes da lista
def imprimirListaEspera():
    if lista_hospital.head is None:
        print("A fila está vazia.")
        return

    atual = lista_hospital.head
    print("\n--- Lista de Espera ---")
    while atual is not None:
        print(f"Cartão: {atual.cor} - Número: {atual.numero}") # [4]
        atual = atual.proximo
    print("-----------------------\n")

# [EXIGÊNCIA DE CÓDIGO 6 de 7]: Remoção e atendimento do primeiro da fila
def atenderPaciente():
    if lista_hospital.head is None:
        print("Não há pacientes para atender.")
        return

    atendido = lista_hospital.head
    print(f"\nChamando paciente do cartão {atendido.cor} número {atendido.numero} para atendimento!") # [4]
    lista_hospital.head = lista_hospital.head.proximo # Remove o primeiro da fila [4]

# [EXIGÊNCIA DE CÓDIGO 7 de 7]: Menu principal do sistema
def menu():
    while True:
        print("1 – Adicionar paciente à fila")
        print("2 – Mostrar pacientes na fila")
        print("3 – Chamar paciente")
        print("4 – Sair")

        opcao = input("Escolha uma opção: ") # [5]

        if opcao == '1':
            inserir()
        elif opcao == '2':
            imprimirListaEspera()
        elif opcao == '3':
            atenderPaciente()
        elif opcao == '4':
            print("Encerrando o programa.")
            break
        else:
            print("Opção inválida! Retornando ao menu.") # [5]

if __name__ == "__main__":
    menu()
```
